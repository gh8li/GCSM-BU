```python
import pandas as pd
import numpy as np
import torch
```

```python
def multi_column_sort(tensor_ls, ascending=None, return_indices=False):
    if ascending is None:
        ascending = [True for _ in range(len(tensor_ls))]
    elif isinstance(ascending, bool):
        ascending = [ascending for _ in range(len(tensor_ls))]
        
    cur_sort_indices = torch.argsort(tensor_ls[-1], stable=True, descending=not ascending[-1])
    for i in range(-2, -len(tensor_ls)-1, -1):
        cur_to_sort = torch.index_select(tensor_ls[i], 0, cur_sort_indices)
        cur_temp_sort_indices = torch.argsort(cur_to_sort, stable=True, descending=not ascending[i])
        cur_sort_indices = cur_sort_indices[cur_temp_sort_indices]
    # sorted_tensor_ls = [t[cur_sort_indices] for t in tensor_ls]

    sorted_tensor_ls = []
    for t in tensor_ls:
        sorted_tensor_ls.append(torch.index_select(t, 0, cur_sort_indices))
    if return_indices:
        return sorted_tensor_ls, cur_sort_indices
    # get_agent_array
    return sorted_tensor_ls
```
